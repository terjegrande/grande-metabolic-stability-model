```python
# --- Imports ---
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("seaborn-whitegrid")

# --- Monte Carlo Core ---
def monte_carlo_simulation(func, samples=10000, low=0, high=1):
    """Kjører Monte Carlo på en funksjon."""
    x = np.random.uniform(low, high, samples)
    y = func(x)
    return x, y, np.mean(y)

# --- Example Function ---
def example_function(x):
    return np.sin(5 * x) + x**2

# --- Run Simulation ---
x, y, mean_value = monte_carlo_simulation(example_function, samples=50000)

# --- Plot ---
plt.figure(figsize=(7,4))
plt.scatter(x, y, s=2, alpha=0.3)
plt.title(f"Monte Carlo Simulation (mean={mean_value:.3f})")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.show()
```
