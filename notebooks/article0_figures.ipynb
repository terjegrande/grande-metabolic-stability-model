```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

# Create figure
fig, ax = plt.subplots(figsize=(8, 7))

# Triangle coordinates
points = np.array([
    [0, 0],      # ζ
    [1, 0],      # L
    [0.5, 0.85]  # S
])

# Draw triangle
triangle = patches.Polygon(points, closed=True, fill=False, linewidth=2)
ax.add_patch(triangle)

# Node positions
zeta = points[0]
lambda_ = points[1]
sigma = points[2]

# Draw nodes
ax.scatter(*zeta, color="blue", s=200)
ax.scatter(*lambda_, color="green", s=200)
ax.scatter(*sigma, color="orange", s=200)

# Labels
ax.text(zeta[0], zeta[1] - 0.06, "ζ (Friksjon)", ha="center", fontsize=14, color="blue")
ax.text(lambda_[0], lambda_[1] - 0.06, "L (Kontroll)", ha="center", fontsize=14, color="green")
ax.text(sigma[0], sigma[1] + 0.04, "S (Stabilitet)", ha="center", fontsize=14, color="orange")

# Arrows between nodes
arrow_style = dict(arrowstyle="->", color="black", linewidth=1.8)

ax.annotate("", xy=sigma, xytext=zeta, arrowprops=arrow_style)
ax.annotate("", xy=lambda_, xytext=sigma, arrowprops=arrow_style)
ax.annotate("", xy=zeta, xytext=lambda_, arrowprops=arrow_style)

# Arrow labels
ax.text((zeta[0] + sigma[0]) / 2 - 0.05, (zeta[1] + sigma[1]) / 2, "Friksjon → Stabilitet", fontsize=10)
ax.text((sigma[0] + lambda_[0]) / 2 + 0.05, (sigma[1] + lambda_[1]) / 2, "Stabilitet → Kontroll", fontsize=10)
ax.text((lambda_[0] + zeta[0]) / 2, (lambda_[1] + zeta[1]) / 2 - 0.05, "Kontroll → Friksjon", fontsize=10)

# Title
plt.title("ζ–L–S Rammeverket for Metabolsk Stabilitet", fontsize=16)

# Clean up axes
ax.set_xlim(-0.1, 1.1)
ax.set_ylim(-0.1, 1.0)
ax.axis("off")

# Save files
plt.savefig("zls_framework.png", dpi=300, bbox_inches="tight")
plt.savefig("zls_framework.svg", dpi=300, bbox_inches="tight")

plt.show()
```
