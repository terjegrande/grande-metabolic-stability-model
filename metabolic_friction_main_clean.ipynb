```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("seaborn-v0_8-whitegrid")
# --- Imports ---
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")

# --- Friction Model ---

def friction_U(x, a=8, x0=0.5, b=0.2):
    """
    U-formet friksjonskurve.
    Minimum ved x0.
    """
    return a * (x - x0)**2 + b


def metabolic_friction(NAD_ratio, homocysteine, weight_nad=0.6, weight_hcy=0.4):
    return weight_hcy * homocysteine + weight_nad * (1 / (NAD_ratio + 1e-9))


def nuclear_friction(DSB_load, chromatin_noise, w_dsb=0.7, w_noise=0.3):
    return w_dsb * DSB_load + w_noise * chromatin_noise


def total_friction(z_metabolic, z_nuclear):
    return z_metabolic + z_nuclear


# --- Aepi Model ---

def SIRT_ledig(SIRT_total, DSB_load, k_dsb=0.15):
    return SIRT_total * np.exp(-k_dsb * DSB_load)


def gamma_friksjon(NAD_ratio, homocysteine, alpha=1.8, beta=0.6):
    return alpha * homocysteine - beta * NAD_ratio


def MGI(NAD_ratio, ATP_ratio, weight_nad=0.65, weight_atp=0.35):
    return weight_nad * NAD_ratio + weight_atp * ATP_ratio


def Aepi(SIRT_ledig_value, gamma_value, MGI_value,
         w_sirt=0.45, w_gamma=0.30, w_mgi=0.25):
    return (w_sirt * SIRT_ledig_value) - (w_gamma * gamma_value) + (w_mgi * MGI_value)


# --- Figure Generator ---

def generate_figure(x, y, title, xlabel, ylabel, filename):
    plt.figure(figsize=(7, 4))
    plt.plot(x, y, linewidth=2)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.tight_layout()
    plt.savefig(filename, dpi=300)
    plt.show()
    print(f"Saved: {filename}")


# --- Example: Figure 2.1 ---

x = np.linspace(0, 1, 500)
y = friction_U(x)

generate_figure(
    x, y,
    title="U-shaped Friction Curve",
    xlabel="x",
    ylabel="Friction U(x)",
    filename="figure_2_1_friction_U.png"
)
```
