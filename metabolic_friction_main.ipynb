```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def friction_U(x, a=8, x0=0.5, b=0.2):
    return a * (x - x0)**2 + b
```

```python
x = np.linspace(0, 1, 300)
F = friction_U(x)

plt.figure(figsize=(6,4))
plt.plot(x, F, linewidth=3)
plt.title("Figure 2.1 – U-shaped Friction Curve")
plt.xlabel("Input level")
plt.ylabel("Friction F(t)")
plt.grid(True)
plt.tight_layout()
plt.show()
```
