```python
# --- Imports ---
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("seaborn-v8-whitegrid")

# --- Substance Simulation Model ---
def simulate_substance_decay(initial_value, k, t):
    """Eksponentiell nedbrytning."""
    return initial_value * np.exp(-k * t)

def simulate_substance_production(rate, t):
    """Lineær produksjon."""
    return rate * t

def combined_substance_model(initial_value, k, rate, t):
    decay = simulate_substance_decay(initial_value, k, t)
    production = simulate_substance_production(rate, t)
    return decay + production

# --- Plot Function ---
def plot_simulation(t, y, title, xlabel, ylabel):
    plt.figure(figsize=(7,4))
    plt.plot(t, y, linewidth=2)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.grid(True)
    plt.show()

# --- Example Run ---
t = np.linspace(0, 10, 200)
y = combined_substance_model(initial_value=10, k=0.3, rate=0.5, t=t)
plot_simulation(t, y, "Substance Simulation", "Time", "Level")
```
